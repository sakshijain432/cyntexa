```python
dbutils.widgets.text("catalog","claims")
catalog = dbutils.widgets.get("catalog")
```

###plans

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.raw.plans AS (
    SELECT * ,
    _metadata.file_path AS src_file,
    current_timestamp() AS ingested_at
    FROM read_files('/Volumes/claims/raw/landing/plans.csv', format => 'csv', header => 'true')
)          
""")
```

```python
spark.sql(f"SELECT COUNT(*) AS record_count FROM {catalog}.bronze.plans").display()
```
